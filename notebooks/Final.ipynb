## **Feature Engineering**

```python
df["Sleep Duration"].value_counts()
```

```python
df.loc[df["Sleep Duration"] == "Others", "Sleep Duration"] = "More than 8 hours by far (16 , 17 hours) in continuous"
```

```python
df["Sleep Duration"].value_counts()
```

```python
df['Risk_Flag'] = ((df['Have you ever had suicidal thoughts ?'] == 'Yes') & (df['Family History of Mental Illness'] == 'Yes')).astype(int)
```

 ## **1. Creating the Risk Flag Feature**

This feature identifies students who have both suicidal thoughts and a family history of mental illness.

A value of 1 means that both conditions are present. A value of 0 means that one or both conditions are not present.

This feature represents a combined mental health risk indicator.

```python
stress_cols = ['Academic Pressure', 'Financial Stress', 'Work/Study Hours']
df['Stress_Index'] = df[stress_cols].mean(axis=1)
```

## **2. Creating the Stress Index Feature**

This feature combines academic pressure, financial stress, and work/study hours into one general stress indicator.

The Stress Index is calculated by taking the average of these three columns for each student.

A higher Stress Index may indicate that the student is facing greater academic, financial, or workload-related stress.

```python
df['Pressure_per_Satisfaction'] = (df['Academic Pressure'] / (df['Study Satisfaction'] + 1))
```

## **3. Creating the Pressure-to-Satisfaction Feature**

This feature compares academic pressure with study satisfaction.

Students with high academic pressure and low study satisfaction will have a higher value. This may indicate that they are struggling to cope with their academic workload.

The value 1 is added to study satisfaction to avoid division by zero

## **Preprocessing and Feature Scaling**

```python
x = df.drop(columns=['Depression'])
y = df['Depression']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
def degree_to_level(degree):
    degree = str(degree).strip().upper()

    if degree in ['CLASS 10', 'CLASS 12']:
        return 0

    elif degree in [
        'B.ARCH', 'B.COM', 'B.ED', 'B.PHARM', 'B.TECH',
        'BA', 'BBA', 'BCA', 'BE', 'BHM', 'BSC',
        'LLB', 'MBBS'
    ]:
        return 1

    elif degree in [
        'LLM', 'M.COM', 'M.ED', 'M.PHARM', 'M.TECH',
        'MA', 'MBA', 'MCA', 'MD', 'ME', 'MHM', 'MSC'
    ]:
        return 2

    elif degree == 'PHD':
        return 3

    else:
        return 4
```

```python
x_train['Degree_Level'] = x_train['Degree'].apply(degree_to_level)
x_test['Degree_Level'] = x_test['Degree'].apply(degree_to_level)
```

```python
x_train.drop(columns=['Degree'], inplace=True)
x_test.drop(columns=['Degree'], inplace=True)
```

```python
binary_mappings = {'Family History of Mental Illness': {'Yes': 1, 'No': 0},'Have you ever had suicidal thoughts ?': {'Yes': 1, 'No': 0},'Gender': {'Male': 1, 'Female': 0}}
for column, mapping in binary_mappings.items():
    x_train[column] = x_train[column].map(mapping)
    x_test[column] = x_test[column].map(mapping)
```

```python
sleep_encoder = OrdinalEncoder(
    categories=[[
        'Less than 5 hours',
        '5-6 hours',
        '7-8 hours',
        'More than 8 hours',
        'More than 8 hours by far (16 , 17 hours) in continuous'
    ]]
)

x_train[['Sleep Duration']] = sleep_encoder.fit_transform(
    x_train[['Sleep Duration']]
)

x_test[['Sleep Duration']] = sleep_encoder.transform(
    x_test[['Sleep Duration']]
)
```

```python
encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore',drop='first')
x_train_encoded = encoder.fit_transform(x_train[['Dietary Habits']])
x_test_encoded = encoder.transform(x_test[['Dietary Habits']])
encoded_columns = encoder.get_feature_names_out(['Dietary Habits'])
x_train_encoded = pd.DataFrame(x_train_encoded,columns=encoded_columns,index=x_train.index)
x_test_encoded = pd.DataFrame(x_test_encoded,columns=encoded_columns,index=x_test.index)
```

```python
x_train.drop(columns=['Dietary Habits'], inplace=True)
x_test.drop(columns=['Dietary Habits'], inplace=True)

x_train = pd.concat([x_train, x_train_encoded], axis=1)
x_test = pd.concat([x_test, x_test_encoded], axis=1)
```

```python
scaler = RobustScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled,columns=x_train.columns,index=x_train.index)
x_train_scaled
```

```python
x_test_scaled = pd.DataFrame(x_test_scaled,columns=x_test.columns,index=x_test.index)
x_test_scaled
```

# **Models Training**

```python
from sklearn.ensemble import GradientBoostingClassifier

models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Support Vector Machine": SVC(probability=True, random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)
}

model_performance = {}

print("=== Starting training of initial models ===")
for model_name, model in models.items():
    print(f"Training {model_name}...")

    model.fit(x_train_scaled, y_train)

    y_pred = model.predict(x_test_scaled)
    y_prob = model.predict_proba(x_test_scaled)[:, 1]

    accuracy = accuracy_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_prob)

    model_performance[model_name] = {
        "Accuracy": accuracy,
        "AUC-ROC": roc_auc
    }

    print(f"---- {model_name} ----")
    print(f"Accuracy: {accuracy:.4f} | AUC-ROC: {roc_auc:.4f}")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    print("="*50)
```

```python
models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Support Vector Machine": SVC(probability=True, random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)
}

print("=== Starting model training ===")
for model_name, model in models.items():
    print(f"Training {model_name}...")
    model.fit(x_train_scaled, y_train)

def plot_confusion_matrix(y_true, y_pred, model_name="Model"):
    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(6, 4.5))
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        cbar=True,
        xticklabels=['No Depression (0)', 'Depression (1)'],
        yticklabels=['No Depression (0)', 'Depression (1)']
    )

    plt.title(f'Confusion Matrix - {model_name}', fontsize=14, pad=15)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.ylabel('Actual Label', fontsize=12)
    plt.tight_layout()
    plt.show()

print("\n=== Plotting confusion matrices for all models ===")
for model_name, model in models.items():
    y_pred = model.predict(x_test_scaled)
    plot_confusion_matrix(y_test, y_pred, model_name=model_name)
```

# **Hyperparameter Tuning Using GridSearchCV**

```python
param_grid_rf = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

print("We are currently searching for the best parameters for the Random Forest model")

grid_search_rf = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid_rf,
    cv=5,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)

grid_search_rf.fit(x_train_scaled, y_train)

best_rf_model = grid_search_rf.best_estimator_

print("\n Random Forest has been successfully completed")
print("The best parameters found:")
print(grid_search_rf.best_params_)

y_pred_rf = best_rf_model.predict(x_test_scaled)
y_prob_rf = best_rf_model.predict_proba(x_test_scaled)[:, 1]

print("\n Final evaluation after tuning (Tuned Random Forest)")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"AUC-ROC: {roc_auc_score(y_test, y_prob_rf):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```
